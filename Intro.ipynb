# Introduction to Python: Output, Input and Variables

**How to use this notebook (read this first!)**
- This page is called a *Jupyter Notebook*. It is made of small boxes called **cells**.
- A **text cell** (like this one) just holds notes for you to read.
- A **code cell** holds instructions for the computer. Click on it and press `Shift + Enter` to run it. The answer appears right below.
- In code cells, any line starting with `#` is a **comment**: a note to humans that the computer completely ignores.

**The three big ideas on this page**
- **Output** : the `print( )` function shows something on the screen. Think of it as the computer *talking to you*.
- **Input** : the `input( )` function asks the person using the program to type something. Think of it as the computer *listening to you*.
- **Variables** : a labelled box (container) used to store information so you can use it later. The label on the box is called the `variable name` (also known as an `identifier`).

> A **function** is just a ready-made mini-tool with a name. You use it by writing its name followed by round brackets `( )`. Whatever you put inside the brackets is what you hand over to the tool to work on.


## Output

Output means *anything the computer shows back to you*. In Python we do that with `print( )`.


```python
# print() is a built-in function that displays whatever is inside the brackets
# "hello world" is text. Text must be wrapped in quotes so Python knows it is text and not a command
# Traditionally "hello world" is the very first thing every programmer prints
print("hello world")
```

```python
# Python is also a powerful calculator. No quotes here, so Python treats these as numbers, not text
# ** means "to the power of", so 34 ** 3 is 34 x 34 x 34 (34 multiplied by itself 3 times)
# * means multiply, so the two big results get multiplied together
# Python does the ** parts first, then the * (just like maths class: powers before multiplying)
# print() then shows the final answer. Python handles huge numbers without any trouble
print(34 ** 3 * 34 ** 33)
```

## Input

Input means *information the user gives to the program*, usually by typing. In Python we do that with `input( )`.


```python
# input() pauses the program and shows a small box where you can type
# The text inside the quotes is the question (called a "prompt") shown to the user
# Press Enter after typing and the program carries on
# Note: whatever you type comes back as TEXT, even if you type digits like 23
# Here we are not saving the answer anywhere, so it is just displayed. (Next up: variables, which can save it!)
input("Enter your name :")
```

## Variables

A variable is like a **labelled box** (container). You put some information inside the box, and stick a label on the outside so you can find it again. The label is the `variable name` or `identifier`. 

Real life example: a jar in your kitchen with the label "sugar". The label is the variable name, the sugar is the value stored in it. You can later swap the sugar for salt, and the label stays the same.

### How to create a variable ?
The moment you assign a value to a word, it becomes a variable. 
- The `=` sign here does **not** mean "equals" like in maths. It means **"store the thing on the right inside the box named on the left"**. That is called *assignment*.
- You don't need to announce or declare it beforehand. Python makes the box for you.


```python
# Make a box labelled "age" and put the number 23 inside it
# Nothing appears on screen because we only stored it, we did not print it
age = 23
```

```python
# Make a box labelled "x" and put the text "john" inside it
# Quotes around john = it is text (called a string). Without quotes Python would hunt for a variable named john
x = "john"
```

```python
# print(age) looks inside the box labelled age and shows what is stored there: 23
# Note: there are NO quotes around age, so Python reads it as a variable name, not as the word "age"
print(age)
# same idea: opens the box labelled x and shows its content: john
print(x)
```

# Variable naming conventions (the rules for naming your boxes)

Python is strict about names. Break a rule and you get an error message. Here are the rules, in plain words:

<ul>
    <li><b>No Space</b> : <code>my age</code> is wrong because Python thinks these are two separate words. Use <code>my_age</code> instead.</li>
    <li><b>A variable name cannot start with a number but can include or end with a number</b> : <code>2cats</code> is wrong, but <code>cat2</code> and <code>c2ats</code> are fine.</li>
    <li><b>A variable name cannot start, include or end with a special symbol except "_" (underscore)</b> : symbols like <code>@ # $ % - !</code> are not allowed. So <code>user-name</code> is wrong, but <code>user_name</code> is fine.</li>
    <li><b>Do Not Use Reserved Keywords</b> : Python already uses certain words for its own special jobs (listed in the table below). You cannot borrow them as your own names. For example, you cannot name a variable <code>if</code> or <code>True</code>.</li>
</ul>

<p><b>Also good to know:</b> names are <b>case sensitive</b>. <code>Age</code>, <code>age</code> and <code>AGE</code> are three different boxes.</p>

### Suggestions (not strict rules, but what good programmers do)
- Start a variable name with a lowercase letter only. 
- Give names that explain themselves: `age` is much better than `a`.
- When a name is made of several words, e.g., `helloworldmynameis` is very hard to read. Two popular fixes:
    - helloWorldMyNameIs : `camelCasing` (every new word starts with a capital letter, humps like a camel's back)
    - hello_world_my_name_is : `snake_casing` (words joined by an underscore, like a snake lying on the ground) 
    - Python programmers usually prefer snake_casing.


# Data Types

Every piece of information in Python has a **type**, which tells Python what *kind* of thing it is and what you are allowed to do with it. For example, you can add two numbers, but adding "apple" and a number makes no sense.


### Primitive Data types
"Primitive" just means the basic building blocks that everything else is made from.

- **int** (number without decimal) : whole numbers, e.g. `7`, `-3`, `0`, `2026`. The word is short for *integer*.
- **float** (Number with decimal) : numbers with a decimal point, e.g. `3.14`, `-0.5`, `99.0`.
- **Bool / Boolean** : `True` / `False`. Only two possible values, like a light switch (on or off). It must be written with a capital T or F. Used for yes/no questions.
- **String(str)** : sequence of characters (alphabets, numbers, special symbols and whitespace) enclosed in single or double quotes, e.g. `'hi'` or `"hi"`. Anything inside quotes is a string, even if it looks like a number.
- **NoneType** : intentional absence of value. `None` is Python's way of saying "there is nothing here on purpose". It is like an empty box with a label on it, which is different from a box holding zero (0) or an empty piece of text.


```python
# type() is a function that tells you what kind of data something is
# print() then shows that answer on the screen. Each line below is checked separately.

# Letters inside quotes: a string, so Python answers <class 'str'> (str = string)
print(type("rohan"))
# Letters, a symbol and digits all inside quotes: still one string
print(type("rohan@123"))
# Looks like a number but it has quotes, so it is TEXT, not a number. Still <class 'str'>
print(type("123"))
# Only spaces inside quotes: a space is a character too, so this is also a string
print(type("   "))
```

```python
# None means "no value on purpose". Here the box named mynameis is created but holds nothing yet
# Notice there are no quotes: None is a special Python word, not text
mynameis = None 
```

```python
# Re-using the same label with a new value: the old content (None) is thrown away
# and the box now holds the text "john doe". A variable can be updated any time
mynameis = "john doe"
```

```python
# Another empty box: yourage exists but has no real value yet. We will fill it in the next cell
yourage = None
```

```python
# Now the box holds the whole number 234 (an int). The earlier None is replaced
yourage = 234
```
