# How to check the data type of a value / variable

`type()` returns the data type of the value / variable specified


#### note : variable doesn't have a type of their own, their type depends upon the value in them

````python
name = "john doe"
print(type(name)) # variable
print(type("monty")) # value
print(type(123)) # value
print(type("123")) # value
print(type("hello@123")) # value
print(type("   ")) # value
````

# String
a sequence of characters `(alphabets, special symbols, numbers and whitespace)` enclosed in single `' '` or double `" "` quotes

### Note : value collected using the input function is of type `str` by default but can be converted into supported data type

````python
name = input("Enter your name : ")
username = input("Enter your username : ")
age = input("Enter age : ")

print(name , type(name))
print(username , type(username))
print(age , type(age))
````

# indexing
counting starting from `0` used to access characters of a string

---

[] are used to use indexing

---


```python
# h e l l o
# 1 2 3 4 5 : normal counting (to get length)
# 0 1 2 3 4 : positive indexing (L -> R)
#-5-4-3-2-1 : negative indexing (R -> L)
```


positive indexing
- first character is always on the `0th` position
- last character is always on the `len(str) - 1` = th position

negative indexing
- first character is always on the -1th position

--- 
##### str[ start : stop : step ]

#### getting a single character


##### str[ start ]

````python
txt = "hello world"
````

````python
print(txt[-2]) # l
print(txt[9]) # l
print(txt[0]) # h
print(txt[-1]) # d
print(txt[ 10 ]) # d 
print(txt[ len(txt) - 1 ]) # d
````

#### Slicing
get a substring

##### str [ start(inclusive) : stop(exclusive) ] 

````python
print(txt[0 : 4])  # 0 : 3
````

````python
print(txt[6 : 11])  # 6 : 10
````

````python
print(txt[6 : ])  # 6 : end
````

````python
print(txt[ : 5])  # start : 4
````

````python
print(txt[-5 : -1]) # -5 : -2
````

````python
print(txt[-5 : ]) # -5 : end
````
