# How to check the data type of a value / variable

`type()` returns the data type of the value / variable specified


#### note : variable doesn't have a type of their own, their type depends upon the value in them

````python
# type() tells you what kind of data something is
name = "john doe"          # quotes = text, so this is a string
print(type(name))          # looks inside the box called name -> str
print(type("monty"))       # text in quotes -> str
print(type(123))           # no quotes, plain number -> int
print(type("123"))         # digits inside quotes are still text -> str
print(type("hello@123"))   # letters, symbols and digits mixed together -> str
print(type("   "))         # spaces are characters too -> str
````

# String
a sequence of characters `(alphabets, special symbols, numbers and whitespace)` enclosed in single `' '` or double `" "` quotes

### Note : value collected using the input function is of type `str` by default but can be converted into supported data type

````python
# input() always hands back text, whatever the user types
name = input("Enter your name : ")
username = input("Enter your username : ")
age = input("Enter age : ")        # typing 23 gives you the text "23", not the number 23

print(name , type(name))           # a comma in print() shows both things, side by side
print(username , type(username))
print(age , type(age))             # says str, even though it looks like a number
````

````python
# same as above, but this time we turn the age into a real number
name = input("Enter your name : ")
username = input("Enter your username : ")
age =  int(input("Enter age : "))  # input() reads the text, int() converts it to a whole number

print(name , type(name))
print(username , type(username))
print(age , type(age))             # now it says int, so we can do maths with it
````

````python
print( age * age)   # multiplying works because age is an int now; text would not multiply like this
````

## Concatenation
adding two or more strings together using the `+` to form a new string

````python
# concatenation: + glues strings together into one bigger string
# watch the spaces inside the quotes, python will not add them for you
print("hello my name is " + name + " and my user name is " + username)
````

````python
# this line breaks on purpose: age is an int and + cannot mix text with a number
# TypeError: can only concatenate str (not "int") to str
# to fix it, wrap it as str(age), or use an f-string like in the next section
print("hello my name is " + name + " and i am a " + age + " year old youtuber and my user name is " + username)
````

# Formatted Strings

strings created using `f""` or `f''`, where values / variables and expressions are inserted in `{}`

````python
# f-string: put an f before the quote, then drop variables straight into { }
# no + to juggle, and numbers need no converting
print(f'hello my name is {name} and i am a {age} year old youtuber and my user name is {username}')
````

# indexing
counting starting from `0` used to access characters of a string

---

[] are used to use indexing

---


```python
# h e l l o
# 1 2 3 4 5 : normal counting (to get length)
# 0 1 2 3 4 : positive indexing (L -> R)
#-5-4-3-2-1 : negative indexing (R -> L)
```


positive indexing
- first character is always on the `0th` position
- last character is always on the `len(str) - 1` = th position

negative indexing
- first character is always on the -1th position

--- 
##### str[ start : stop : step ]

#### getting a single character


##### str[ start ]

````python
txt = "hello world"   # 11 characters, the space in the middle is one of them
````

````python
# txt[ position ] pulls out one single character
print(txt[-2]) # l  -> second from the right
print(txt[9]) # l   -> counting from the left, where the first one is 0
print(txt[0]) # h   -> first character is always 0, never 1
print(txt[-1]) # d  -> last character, the short way
print(txt[ 10 ]) # d  -> last character the long way: 11 characters, so the biggest index is 10
print(txt[ len(txt) - 1 ]) # d  -> same answer worked out, and it fits a string of any length
# ask for txt[11] and python stops with IndexError, there is nothing there
````

#### Slicing
get a substring

##### str [ start(inclusive) : stop(exclusive) ] 

````python
# slicing: txt[start : stop] keeps start, leaves out stop
print(txt[0 : 4])  # 0 : 3  -> hell, so position 4 (the o) is dropped
````

````python
print(txt[6 : 11])  # 6 : 10  -> world, stop is allowed to sit one past the last index
````

````python
print(txt[6 : ])  # 6 : end  -> leave stop empty and it runs to the end
````

````python
print(txt[ : 5])  # start : 4  -> leave start empty and it begins at 0
````

````python
print(txt[-5 : -1]) # -5 : -2  -> worl, negative numbers slice the same way
````

````python
print(txt[-5 : ]) # -5 : end  -> world, an easy way to grab the last 5 characters
````
